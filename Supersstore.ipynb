```python
import pandas as pd ;
```

```python
data = pd.read_csv('Week4-DataAnalytics/Superstore.csv', encoding='latin1')
data.head(10)
```

```python
data.info()
```

```python
data['Order Date'] = pd.to_datetime(data["Order Date"])
data['Ship Date'] = pd.to_datetime(data['Ship Date'])
```

```python
data.info()
```

```python

pd.set_option('display.float_format', '{:,.2f}'.format)


total_sales = data['Sales'].sum()
total_profit = data['Profit'].sum()
total_orders = data['Order ID'].nunique()
total_quantity = data['Quantity'].sum()
profit_margin = (total_profit / total_sales) * 100
avg_order_value = total_sales / total_orders
avg_discount = data['Discount'].mean() * 100

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Quantity Sold: {total_quantity:,}")
print(f"Profit Margin: {profit_margin:.2f}%")
print(f"Average Order Value: ${avg_order_value:,.2f}")
print(f"Average Discount: {avg_discount:.2f}%")
```

```python
sales_by_category = data.groupby('Category')['Sales'].sum().sort_values(ascending=False)
sales_by_subcategory = data.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)
sales_by_region = data.groupby('Region')['Sales'].sum().sort_values(ascending=False)
sales_by_segment = data.groupby('Segment')['Sales'].sum().sort_values(ascending=False)

profit_by_category = data.groupby('Category')['Profit'].sum().sort_values(ascending=False)
profit_by_subcategory = data.groupby('Sub-Category')['Profit'].sum().sort_values(ascending=False)

top_customers = data.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False).head(10)
top_products = data.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)


loss_makers = profit_by_subcategory[profit_by_subcategory < 0]
```

```python
print("=== Sales by Category ===")
print(sales_by_category)

print("\n=== Sales by Sub-Category ===")
print(sales_by_subcategory)

print("\n=== Sales by Region ===")
print(sales_by_region)

print("\n=== Sales by Segment ===")
print(sales_by_segment)

print("\n=== Profit by Category ===")
print(profit_by_category)

print("\n=== Profit by Sub-Category ===")
print(profit_by_subcategory)

print("\n=== Top 10 Customers by Sales ===")
print(top_customers)

print("\n=== Top 10 Products by Sales ===")
print(top_products)

print("\n=== Loss-Making Sub-Categories ===")
print(loss_makers)
```

```python

import matplotlib.pyplot as plt

data['Order Date'] = pd.to_datetime(data['Order Date'])
data['Order Month'] = data['Order Date'].dt.to_period('M')

monthly_sales = data.groupby('Order Month')['Sales'].sum()
```

```python

import seaborn as sns

sns.set_style("whitegrid")

# 1. Monthly sales trend
plt.figure(figsize=(12,5))
monthly_sales.plot(kind='line', marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 2. Sales by category
plt.figure(figsize=(8,5))
sales_by_category.plot(kind='bar', color='steelblue')
plt.title('Sales by Category')
plt.ylabel('Sales ($)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# 3. Profit by sub-category (highlight losses)
plt.figure(figsize=(10,6))
colors = ['crimson' if x < 0 else 'seagreen' for x in profit_by_subcategory.values]
profit_by_subcategory.plot(kind='barh', color=colors)
plt.title('Profit by Sub-Category')
plt.xlabel('Profit ($)')
plt.tight_layout()
plt.show()

# 4. Sales by region (pie)
plt.figure(figsize=(7,7))
sales_by_region.plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Sales Share by Region')
plt.ylabel('')
plt.tight_layout()
plt.show()

# 5. Segment comparison — sales vs profit
fig, ax = plt.subplots(figsize=(8,5))
segment_summary = data.groupby('Segment')[['Sales','Profit']].sum()
segment_summary.plot(kind='bar', ax=ax)
plt.title('Sales vs Profit by Segment')
plt.ylabel('$')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# 6. Discount vs Profit relationship (scatter)
plt.figure(figsize=(8,5))
sns.scatterplot(data=data, x='Discount', y='Profit', alpha=0.4)
plt.title('Discount vs Profit')
plt.tight_layout()
plt.show()

# 7. Top 10 products by sales
plt.figure(figsize=(10,6))
top_products.plot(kind='barh', color='darkorange')
plt.title('Top 10 Products by Sales')
plt.xlabel('Sales ($)')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
